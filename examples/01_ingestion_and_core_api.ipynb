# In-Memory Analytics

Supported inputs are local paths, `file://` URIs, supported cloud/object URIs, and non-recursive directories containing JSON/JSONL/XML/CSV/Parquet files. Analytical `to_*` functions return a `Result`; `result.clean_data` contains the requested analytical object and `result.stats` reports row counts.

```python
import json
from pathlib import Path

import schema_sanitizer as ss

out_dir = Path("files/01_ingestion_and_core_api/exercise_01_input_sources")
out_dir.mkdir(parents=True, exist_ok=True)
jsonl_path = out_dir / "events.jsonl"
csv_path = out_dir / "events.csv"
rows = [
    {"id": 1, "active": "yes", "score": "10.5"},
    {"id": 2, "active": "no", "score": "11.0"},
]
jsonl_path.write_text("".join(json.dumps(r) + "\n" for r in rows), encoding="utf-8")
csv_path.write_text("id,active,score\n1,yes,10.5\n2,no,11.0\n", encoding="utf-8")
```

```python
file_result = ss.to_pyarrow(jsonl_path, input_format="jsonl")
print(file_result.clean_data.to_pylist())
print(file_result.clean_data.schema)
```

```python
result = ss.to_pyarrow(csv_path, input_format="csv", true_tokens=("yes",), false_tokens=("no",))
print(result.clean_data.schema)
print(result.stats)
```
