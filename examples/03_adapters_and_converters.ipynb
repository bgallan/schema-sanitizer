# Adapters and Converters

Analytical `to_*` functions return a `Result` whose `clean_data` is PyArrow, pandas, Polars, or DuckDB. File-to-file `to_*` functions stream sanitized output to CSV, JSON Lines, or Parquet.

```python
from pathlib import Path

import schema_sanitizer as ss

out_dir = Path('files/03_adapters_and_converters/exercise_01_converters')
out_dir.mkdir(parents=True, exist_ok=True)
path = out_dir / 'events.jsonl'
path.write_text('{"id": 1, "country": "ES"}\n{"id": 2, "country": "US"}\n', encoding='utf-8')
```

```python
arrow_result = ss.to_pyarrow(path, input_format='jsonl')
result = ss.to_pyarrow(path, input_format='jsonl')
print(arrow_result.clean_data.to_pylist())
print(result.stats)
```

```python
# Optional adapters; skip cells if the dependency is not installed.
try:
    pandas_df = ss.to_pandas(path, input_format='jsonl').clean_data
    print(pandas_df)
except ImportError as exc:
    print(exc)
try:
    polars_df = ss.to_polars(path, input_format='jsonl').clean_data
    print(polars_df)
except ImportError as exc:
    print(exc)
try:
    duckdb_rel = ss.to_duckdb(path, input_format='jsonl').clean_data
    print(duckdb_rel.fetchall())
except ImportError as exc:
    print(exc)
```

```python
csv_result = ss.to_csv(path, out_dir / 'events.csv')
jsonl_result = ss.to_jsonl(path, out_dir / 'events.jsonl')
ss.to_jsonl(path, out_dir / 'events.ndjson')
parquet_result = ss.to_parquet(path, out_dir / 'events.parquet')
{'clean_data': (csv_result.clean_data, jsonl_result.clean_data, parquet_result.clean_data), 'files': [p.name for p in out_dir.iterdir()]}
```
