# Large CSV to Parquet

File-to-file converters use bounded output batching. Schema inference scans the input before streaming sanitized output.

```python
from pathlib import Path

import schema_sanitizer as ss

out_dir = Path('files/04_streaming_large_csv_to_parquet/exercise_01_large_csv_stress')
out_dir.mkdir(parents=True, exist_ok=True)
csv_path = out_dir / 'source_large.csv'
parquet_path = out_dir / 'cleaned.parquet'
rows = 10000
with csv_path.open('w', encoding='utf-8') as f:
    f.write('id,value\n')
    for i in range(rows):
        f.write(f'{i},value_{i}\n')
```

```python
result = ss.to_parquet(
    csv_path,
    parquet_path,
    input_format='csv',
    memory_limit_bytes=256 * 1024 * 1024,
)
print({'exists': parquet_path.exists(), 'clean_data': result.clean_data, 'stats': result.stats})
```
